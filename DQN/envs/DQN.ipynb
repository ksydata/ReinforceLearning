```python
import gymnasium as gym
import numpy as np
import cv2 as cv
import ale_py
```

```python
def play_Pong(env):
    s, info = env.reset()
    while True:
        a = np.random.choice([0, 2, 3], 1, p=[1/3, 1/3, 1/3])[0]  # 정지(0), 상(2), 하(3)
        s, r, terminated, truncated, info = env.step(a)

        cv.imshow('Pong animation', cv.cvtColor(env.render(), cv.COLOR_BGR2RGB))
        # cv.waitKey(20)
        cv.waitKey(1)
        
        if terminated or truncated:
            break

env = gym.make('ALE/Pong-v5', render_mode='rgb_array')
print(env.observation_space, env.action_space)
play_Pong(env)
env.close()

if cv.waitKey() == ord('q'):
    cv.destroyAllWindows()
# Box(0, 255, (210, 160, 3), uint8) Discrete(6)
```
