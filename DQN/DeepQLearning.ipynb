```python
import numpy as np
# 행렬 연산을 위한 라이브러리 로드
# pip install numpy==1.23.0

from dezero import Variable
# 딥러닝 프레임워크에서 넘파이의 다차원 배열(ndarray / tensor)을 감싸는 클래스 로드
# DeZero 0.0.13과 NumPy 2.0.2의 호환성 문제
# AttributeError: module 'numpy' has no attribute 'int'.
import dezero.functions as F 
# 벡터의 내적과 행렬곱 등을 계산하는 matmul() 함수 사용을 위해 로드
```

```python
x_np = np.array(5.0)
# 넘파이의 int 타입으로 5를 표현한 배열 생성
x = Variable(x_np)
# 디제로의 Variable 객체로 변환, 딥러닝 프레임워크에서 사용하기 위해 넘파이 배열을 감싸기 
print(x.data)

y = 3 * x**2
# 넘파이 다차원 배열처럼 사용
print(y.data)
# 연산을 수행한 결과, 3 * (5.0^2) 계산
y.backward()
# Varaible 인스턴스에서 backward() 메서드를 호출
# 역전파(backpropagation) 수행하고 각 변수의 기울기(gradient) 계산, 미분
# $\frac{dy}{dx} = 6x
print(x.grad)
```

```python
def rosenbrock(x_0, x_1):
    """
    로젠브록 함수(Rosenbrock function)를 통해 최솟값이 되게 하는 인수(입력) x_0, x_1을 찾아, 
    최적화 문제 풀이, (x_0, x_1) = (1, 1)
    """
    y = 100 * (x_1 - x_0**2)**2 + (x_0 - 1)**2
    # $y = 100(x_1 - x_0)^2 + (x_0 - 1)^2$
    return y
```

```python
x_0 = Variable(np.array(0.0)) 
# Variable 클래스의 인스턴스 x_0을 생성, 초기값 0.0으로 설정 
x_1 = Variable(np.array(2.0)) 
# Variable 클래스의 인스턴스 x_1을 생성, 초기값 2.0으로 설정

y = rosenbrock(x_0, x_1) 
# rosenbrock() 함수를 호출하여 y 계산, x_0과 x_1을 입력으로 사용
y.backward() 
# 역전파 수행하여 x_0과 x_1의 기울기 계산
```

```python
print(x_0.grad, x_1.grad) 
# x_0과 x_1의 기울기 출력, 최적화 문제에서 각 변수의 변화량을 나타냄
# 이 두 미분값(-2, 400)을 모아 만든 벡터를 기울기 벡터라고 하며, 
# 기울기는 각 지점에서 함수의 출력을 가장 크게 증가시키는 방향을 가리킴
# 즉, 기울기에 마이너스를 곱한 (2, -400) 방향은 y의 값을 가장 크게 감소시키는 방향
```

```python
n_iters = 10000
# 최적화 문제에서 반복 실행 횟수 설정
learningRate = 0.001 
# 학습률(learning rate) 설정, 기울기에 곱하여 변수의 값을 업데이트할 때 사용
x_0 = Variable(np.array(0.0)) 
# Variable 클래스의 인스턴스 x_0을 생성, 초기값 0.0으로 설정 
x_1 = Variable(np.array(2.0)) 
# Variable 클래스의 인스턴스 x_1을 생성, 초기값 2.0으로 설정
```

```python
for i in range(n_iters):
    """최적화 문제에서 반복 실행되는 루프"""
    print(x_0, x_1)
    Y = rosenbrock(x_0, x_1)
    # 로젠브록 메서드로 x_0과 x_1을 입력으로 사용하여 Y 계산

    x_0.cleargrad()
    x_1.cleargrad()
    # 이전 반복에서 더해진 각 변수의 기울기를 초기화

    Y.backward()
    # 미분(역전파) 수행하여 x_0과 x_1의 기울기 계산

    x_0.data -= learningRate * x_0.grad.data
    x_1.data -= learningRate * x_1.grad.data
    # 기울기에 학습률을 곱하여 변수의 값을 업데이트, 
    # 최적화 문제에서 각 변수의 값을 조정하여 함수의 출력을 최소화하는 방향으로 이동
```

- ML, 선형회귀로 x와 y의 선형관계에서 x의 값을 목표로 보고, y의 값을 예측하는 모델(수식)
- 데이터와 예측값의 차이, 잔차(residual)을 최소화해야 하고, 이중에 손실함수(모델의 '나쁨'을 평가) 중 평균제곱오차(MSE)를 사용
- 즉, 손실함수가 최대가 되는 W와 b를 찾아야 하는 함수 최적화 문제이며, 경사 하강법으로 최소화하는 매개변수 탐색

```python
np.random.seed(0)
# 난수 생성 시드, 동일한 난수 시퀀스를 재현하기 위해 설정
x = np.random.rand(100, 1)
# 선형 회귀 문제에서 입력 데이터 x와 타깃 데이터 y 생성
y = 5 + 2*x + np.random.rand(100, 1)
# y = 5+2x + 노이즈, 선형 관계에 약간의 무작위성을 추가하여 데이터 생성

# [비선형 데이터셋 가정]
# y = np.sin(2*np.pi* x) + np.random.rand(100, 1)
# y = F.matmul(x, W) + b

x, y = Variable(x), Variable(y)
W = Variable(np.random.rand(1, 1))
# 선형 회귀 모형의 가중치 W를 Variable로 초기화, 난수로 시작
b = Variable(np.random.rand(1))
# 선형 회귀 모형의 편향 b를 Variable로 초기화, 난수로 시작
```

```python
def predict(x):
    """선형 회귀 모형의 예측 함수"""
    y = F.matmul(x, W) + b
    # 입력 x에 가중치 W를 곱하고 편향 b를 더하여 예측값 계산
    return y

def meanSquaredError(x_0, x_1):
    """평균 제곱 오차(Mean Squared Error) 손실 함수"""
    diff = x_0 - x_1
    return F.mean(F.sum(diff ** 2, axis = 1))
    # F.mean(F.sum((x_0 - x_1)**2, axis=1))
    # 예측값 y와 타깃값 t의 차이를 제곱하여 평균을 계산, 회귀 문제에서 모델의 성능을 평가하는 데 사용
```

```python
learningRate = 0.1 # 학습률 설정, 선형 회귀 모형의 가중치와 편향을 업데이트할 때 사용
n_iters = 100 # 선형 회귀 모형의 학습을 위한 반복 실행 횟수 설정
```

```python
for i in range(n_iters):
    """선형 회귀 모형의 학습을 위한 반복 실행되는 루프"""
    y_pred = predict(x)
    # 현재 가중치 W와 편향 b를 사용하여 입력 x에 대한 예측값 계산

    loss = meanSquaredError(y_pred, y)
    # 예측값과 타깃값 y 사이의 평균 제곱 오차 계산, 모델의 성능을 평가하는 손실 함수

    W.cleargrad()
    b.cleargrad()
    # 이전 반복에서 더해진 가중치 W와 편향 b의 기울기를 초기화

    loss.backward()
    # 미분(역전파) 수행하여 가중치 W와 편향 b의 기울기 계산

    W.data -= learningRate * W.grad.data
    b.data -= learningRate * b.grad.data
    # 기울기에 학습률을 곱하여 가중치 W와 편향 b의 값을 업데이트, 
    # 선형 회귀 모형이 타깃값에 더 가까워지도록 조정

    if i % 10 == 0:
        # 10회마다 현재 반복횟수와 손실값 출력
        print("=======================")
        print(f"iter: {i}, loss: {loss.data}")
        print("W:", W.data, "b:", b.data)
        print("=======================")
```
